# Load packages

```python
import sklearn.metrics as sk
import scanpy as sc
import pandas as pd
import numpy as np
import os
import seaborn as sns
import matplotlib.pyplot as plt
import sklearn.metrics as sm 
from scipy.stats import kendalltau
import seaborn as sns
import matplotlib
```

```python
oripath=r'../../figures/SVF/'
results=os.listdir(oripath)
```

We read all results and format the results into a common framework

```python
results=results[1:]
datasets=[]
algorithm=[]
for d in results:
    datasets.append(d.split('__')[0])
    algorithm.append(d.split('__')[1])
    
unidatasets=np.unique(datasets)
unialgorithm=np.unique(algorithm)
present=pd.DataFrame(index=unidatasets,columns=unialgorithm)
perc_svg_df=pd.DataFrame(index=unidatasets,columns=unialgorithm)
perc_control_svg=pd.DataFrame(index=unidatasets,columns=unialgorithm)

for dataset in unidatasets:
    allres=[]
    for d in results:
        dt=d.split('__')[0]
        alg=d.split('__')[1]
        present.loc[dt,alg]=1
```

```python
def get_svg(ddt):
    potential_svg_cols=['SVG','spatially_variable','highly_variable']
    ddtf=ddt.loc[:,ddt.columns.isin(potential_svg_cols)]
    if ddtf.shape[1]==0:
        pvals_cols=['FDR']
        ddtf=ddt.loc[:,ddt.columns.isin(pvals_cols)]
        ddtf['SVG']=ddtf['FDR']<0.05
        ddtf=ddtf.loc[:,['SVG']]
        if ddtf.shape[1]==0:
            print(ddt.columns)
    ddtf.columns=['SVG']
    return ddtf
```

```python
concatres=pd.concat(allres,axis=1)
concatres=concatres.fillna(False)

for dataset in unidatasets:
    print(dataset)
    allres=[]
    for d in results:
        dt=d.split('__')[0]
        alg=d.split('__')[1]
        if dt==dataset:
            ddt=pd.read_csv(oripath+d,index_col=0)
            ddtf=get_svg(ddt)
            
            perc_svg=np.sum(ddtf['SVG'])/len(ddtf['SVG'])
            perc_svg_df.loc[dt,alg]=perc_svg
            ddtf['control']=[('BLANK' in ind) or ('Neg' in ind) for ind in ddtf.index]
            ddtfcon=ddtf[ddtf['control']==True]
            if ddtfcon.shape[0]>0:
                perc_con_svg=np.sum(ddtfcon['SVG'])/len(ddtfcon['SVG'])
                perc_control_svg.loc[dt,alg]=perc_con_svg
            if alg=='spatialde.csv':
                subi=ddtf.copy()
                subi['gene']=subi.index
                sortedg=subi.groupby('gene').count().sort_values(by='SVG')
                selg=sortedg.index[sortedg.iloc[:,0]==1]
                ddtf=ddtf.loc[ddtf.index.isin(selg),:]
            ddtfsorted=ddtf.sort_index()
            newly=pd.DataFrame(ddtfsorted['SVG'])
            newly.columns=[alg]
            newly=newly.fillna(False)
            allres.append(newly)
    concatres=pd.concat(allres,axis=1)
    concatres=concatres.fillna(False)
    concatres=concatres*1
    try:
        concatres.loc[:,'sinfonia']=concatres.loc[:,'sinfonia_moranI.csv']+concatres.loc[:,'sinfonia_gearyC.csv']>1
        concatres.loc[:,'sinfonia5000']=concatres.loc[:,'sinfonia5000_moranI.csv']+concatres.loc[:,'sinfonia5000_gearyC.csv']>1
    except:
        ee=22
    concatres=concatres.loc[:,~concatres.columns.isin(['sinfonia_moranI.csv','sinfonia_gearyC.csv','sinfonia5000_moranI.csv','sinfonia5000_gearyC.csv'])]
    concatres=concatres.fillna(False)
    concatres=concatres*1
    concatres=concatres.loc[[('BLANK' not in ind) and ('Neg' not in ind) for ind in concatres.index],:]
    sns.clustermap(concatres.astype(float),cmap='Blues')
    jmd=pd.DataFrame(index=concatres.columns,columns=concatres.columns)
    for ind in concatres.columns:
        for col in concatres.columns:
            jmd.loc[ind,col]=sm.jaccard_score(list(concatres.loc[:,ind]),list(concatres.loc[:,col]),average='binary')
    sns.clustermap(jmd.astype(float),figsize=(10,10))
    try:
        for indi in jmd.index:
            for coli in jmd.columns:
                jmdall.loc[indi,coli]=jmdall.loc[indi,coli]+jmd.loc[indi,coli]
    except:
        jmdall=jmd
```

```python
jcc=jmdall.copy()
maxvals={}
for indi in jmdall.index:
    maxvals[indi]=jmdall.loc[indi,indi]
for indi in jmdall.index:
    for coli in jmdall.columns:
        minv=np.min([maxvals[indi],maxvals[coli]])
        jcc.loc[indi,coli]=jmdall.loc[indi,coli]/minv
```

For every dataset, plot the SVGs identified and their scores by each method to explore consistency

```python
overallscore=pd.DataFrame(index=unialgorithm,columns=unialgorithm)
overallscorenum=pd.DataFrame(index=unialgorithm,columns=unialgorithm)
overallscore=overallscore.fillna(0)
overallscorenum=overallscorenum.fillna(0)
import sklearn.metrics as sm 
for dataset in unidatasets:
    allres=[]
    for d in results:
        dt=d.split('__')[0]
        alg=d.split('__')[1]
        if dt==dataset:
            ddt=pd.read_csv(oripath+d,index_col=0)
            if 'seurat' in alg:
                ddt.index=[s.replace('K-','K_') for s in ddt.index]
                ddt.index=[s.replace('e-','e_') for s in ddt.index]
                ddt.index=[s.replace('Y-','Y_') for s in ddt.index]
                ddt.index=[s.replace('word-','word_') for s in ddt.index]
            if 'seurat_markvariogram' in alg:
                ddt['rank']=ddt['rstat'].rank(ascending=False)
            if 'seurat_moransi' in alg:
                ddt['rank']=ddt['pvalue'].rank(ascending=False)
           # if 'squidpy5000_gearyC' in alg:
           #     ddt['rank']=ddt['Pval'].rank(ascending=False)
           # if 'squidpy_gearyC' in alg:
           #     ddt['rank']=ddt['Pval'].rank(ascending=False)
            #compute amount of svg 
            ddtf=get_svg(ddt)
            perc_svg=np.sum(ddtf['SVG'])/len(ddtf['SVG'])
            perc_svg_df.loc[dt,alg]=perc_svg
            # test for blank genes
            ddtf['control']=[('BLANK' in ind) or ('Neg' in ind) for ind in ddtf.index]
            ddtfcon=ddtf[ddtf['control']==True]
            if ddtfcon.shape[0]>0:
                perc_con_svg=np.sum(ddtfcon['SVG'])/len(ddtfcon['SVG'])
                perc_control_svg.loc[dt,alg]=perc_con_svg
            # test for similarity in rank
            ddta=ddt.loc[:,['rank']]
            ddta.columns=[alg]
            present.loc[dataset,alg]=1
            if alg=='spatialde.csv':
                subi=ddta.copy()
                subi['gene']=subi.index
                sortedg=subi.groupby('gene').count().sort_values(by='spatialde.csv')
                selg=sortedg.index[sortedg.iloc[:,0]==1]
                ddta=ddta.loc[ddta.index.isin(selg),:]
            allres.append(ddta)
    concatenates=pd.concat(allres,axis=1)
    concatenates=concatenates.fillna(0)
    sele=pd.DataFrame(np.sum(concatenates,axis=1)).sort_values(by=0,ascending=False)
    concatenates=concatenates.loc[sele.index,:]
    
    plt.figure()
    import seaborn as sns
    import matplotlib
    matplotlib.rcParams['pdf.fonttype'] = 42
    matplotlib.rcParams['ps.fonttype'] = 42
    plt.figure(figsize=(5,5))
    concatenates2=concatenates.copy()
    concatenates2.columns=[indi.replace('.csv','') for indi in concatenates2.columns]
    sns.heatmap(concatenates2,cmap='Purples')
    plt.title(dataset)
    plt.savefig('../../figures/7.spatial_architecture/order_heatmap_svf_'+dataset+'.pdf')
    plt.show()
    pvlsres=pd.DataFrame(index=concatenates.columns,columns=concatenates.columns)
    for ind in concatenates.columns:
        for col in concatenates.columns:
            pvlsres.loc[ind,col]=kendalltau(concatenates.loc[:,ind],concatenates.loc[:,col]).correlation
            overallscore.loc[ind,col]=overallscore.loc[ind,col]+pvlsres.loc[ind,col]
            overallscorenum.loc[ind,col]=overallscorenum.loc[ind,col]+1
    sns.clustermap(pvlsres.astype(float),cmap='Purples',figsize=(5,5))
```
