# Aim of the notebook

This notebook is used to identify domains using **SPACEL**

## Import packages

```python
from SPACEL.setting import set_environ_seed
set_environ_seed(42)
from SPACEL import Splane
import scanpy as sc
import matplotlib
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## Read adata

```python
saving_path=r'../../data/processed_adata/'
adata=sc.read('../../data/processed_adata/adata_multisection_nuclei_r1_with_annotations.h5ad')
adata.X=adata.layers['raw']
adata.obsm['spatial']=np.array(adata.obs.loc[:,['x_centroid','y_centroid']])
```

# ROI1

```python
adata2=adata[(adata.obs['x_centroid']<5000)&(adata.obs['x_centroid']>2000)&(adata.obs['y_centroid']>4000)]
plt.scatter(adata.obs.loc[:,'x_centroid'],adata.obs.loc[:,'y_centroid'],s=0.1)
plt.scatter(adata2.obs.loc[:,'x_centroid'],adata2.obs.loc[:,'y_centroid'],s=0.1)
adata=adata2
Splane.utils.add_cell_type_composition(adata, celltype_anno=adata.obs['Class'])
st_ad_list = [adata]
ndoms=[]
for num in range(1,5):
    ndoms.append(len(np.unique(adata.obs['region_level'+str(num)])))
nclust=ndoms
ksels=[1,2,3]
gnn_dropoutsels=[0.4,0.6,0.8,1]
for gnn_dropoutsel in gnn_dropoutsels: 
    for ksel in ksels:
        for clust in nclust:
            splane_model = Splane.init_model(st_ad_list, n_clusters=clust, k=ksel, gnn_dropout= gnn_dropoutsel, use_gpu=False)
            splane_model.train(d_l=0.2)
            splane_model.identify_spatial_domain()
            new_adata=st_ad_list[0].copy()
            results=new_adata.obs.loc[:,['cell_id','spatial_domain']]
            sc.pl.spatial(st_ad_list[0],color='spatial_domain',spot_size=20)
            import os
            if not os.path.exists('../../figures/7.spatial_architecture/domains_predicted/ROI1/'+str(clust)+'/'):
                os.mkdir('../../figures/7.spatial_architecture/domains_predicted/ROI1/'+str(clust))
            results.to_csv('../../figures/7.spatial_architecture/domains_predicted/ROI1/'+str(clust)+'/predicted_SPACEL_k'+str(ksel)+'gnndropout'+str(gnn_dropoutsel)+'.csv')
```
